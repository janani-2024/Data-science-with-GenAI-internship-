problem 1 : Defanging an IP address

```python
import numpy as np
def defang(ip):
    arr=np.array(list(ip),dtype=object)
    arr[arr=='.']='[.]'
    return"".join(arr)
print(defang("1.1.1.1"))
print(defang("22.45.34.88.10"))
```

problem 2 : Find Numbers with Even number of Digits

```python
def findEven(nums):
    arr = np.array(nums, dtype=str)
    lengths = np.char.str_len(arr)
    return np.sum(lengths % 2 == 0)
print(findEven([11,234,56,90,8]))
```

problem 3 :Number of Good Pairs

```python
def goodpairs(nums):
    nums = np.sort(nums)
    count = 0
    for i in range(len(nums)):
        count += np.sum(nums[i+1:] == nums[i])
    return int(count)
print(goodpairs([1,2,3,1,1,3]))
print(goodpairs([1,1,1,1,1]))
```

problem 4 :How many Numbers are Smaller than the Current Number

```python
def smallernums(nums):
    nums=np.array(nums)
    result=[]
    for i in range(len(nums)):
        count=0
        for j in range(len(nums)):
            if nums[j]<nums[i]:
                count+=1
        result.append(count)
    return result
print(smallernums([8,1,2,2,3]))
```

problem 5 : Substract the Product and Sum of Digits of an Integer

```python
def subtract_product_and_sum(n):
    digits = np.array([int(d) for d in str(n)])
    product_val = np.prod(digits)
    sum_val = np.sum(digits)
    return product_val - sum_val
print(subtract_product_and_sum(234))    
```

problem 6: XOR Operation in an Array

```python
def XORoperation(n, start):
    nums = start + 2 * np.arange(n)
    result = np.bitwise_xor.reduce(nums)
    return int(result)
print(XORoperation(5,0))
print(XORoperation(4,3))
```
